```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, count, round

spark = SparkSession.builder.appName("BookMatch").getOrCreate()

ratings = spark.read.csv(
    "/Volumes/workspace/default/bookmatch/Ratings_clean.csv",
    header=True,
    inferSchema=True
)

print(f"Total records: {ratings.count()}")
ratings.show(5)
```

```python
anomalies = ratings.filter(
    (col("Rating") < 1) | (col("Rating") > 10)
)
print(f"Anomalies outside 1-10: {anomalies.count()}")
print(f"Null Ratings: {ratings.filter(col('Rating').isNull()).count()}")
print(f"Null User_IDs: {ratings.filter(col('User-ID').isNull()).count()}")
```

```python
spark_total = ratings.count()
pandas_total = 433671

spark_avg = ratings.selectExpr("round(avg(Rating),2) as Avg").collect()[0][0]

print(f"PySpark total rows:  {spark_total}")
print(f"Pandas total rows:   {pandas_total}")
print(f"Match: {spark_total == pandas_total}")
print(f"PySpark avg rating:  {spark_avg}")
print(f"Expected avg rating: 7.6")
```
